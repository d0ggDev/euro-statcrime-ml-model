```python
import pandas as pd

# Load the dataset
file_path = "/content/crim_off_cat$defaultview_linear_2_0_cleaned.csv"
df = pd.read_csv(file_path)

# Display the dataset info and preview the first few rows
print("Dataset Information:")
df.info()

print("\nFirst 5 rows:")
display(df.head())
```

```python
for col in df.select_dtypes(exclude=['float64']).columns:
    unique_vals = df[col].unique()
    print(f"Unique values in column '{col}' (count: {len(unique_vals)}):")
    print(unique_vals)
    print("-" * 50)
```

```python
# Identify columns with only one unique value
single_value_cols = [col for col in df.columns if df[col].nunique() == 1]

# Combine with other columns requested for removal
cols_to_remove = list(set(single_value_cols + ['iccs', 'unit', 'geo']))

# Drop the columns
df_cleaned = df.drop(columns=cols_to_remove)

# Split the dataset based on unit_of_measure
df_count = df_cleaned[df_cleaned['unit_of_measure'] == 'Number'].copy()
df_rate = df_cleaned[df_cleaned['unit_of_measure'] == 'Per hundred thousand inhabitants'].copy()

# Remove the unit_of_measure column from the new datasets
df_count = df_count.drop(columns=['unit_of_measure'])
df_rate = df_rate.drop(columns=['unit_of_measure'])

# Save the new datasets to CSV files
df_count.to_csv('/content/crime_data_counts.csv', index=False)
df_rate.to_csv('/content/crime_data_rates.csv', index=False)

# Display status and shapes
print("Saved new datasets:")
print(f"- Counts data (df_count): /content/crime_data_counts.csv (Shape: {df_count.shape})")
print(f"- Rates data (df_rate): /content/crime_data_rates.csv (Shape: {df_rate.shape})")

print("\nPreview of Counts Dataset:")
display(df_count.head())

print("\nPreview of Rates Dataset:")
display(df_rate.head())
```

```python
# Filter the counts dataset for the relevant crime categories
categories = ['Sexual violence', 'Rape', 'Sexual assault']
df_filtered = df_count[df_count['international_classification_of_crime_for_statistical_purposes_iccs'].isin(categories)]

# Pivot the data to align categories side-by-side for comparison by country and year
pivot_df = df_filtered.pivot_table(
    index=['geopolitical_entity_reporting', 'time_period'],
    columns='international_classification_of_crime_for_statistical_purposes_iccs',
    values='obs_value'
).reset_index()

# Check if Rape + Sexual assault equals Sexual violence or how they relate
if 'Sexual violence' in pivot_df.columns and 'Rape' in pivot_df.columns and 'Sexual assault' in pivot_df.columns:
    pivot_df['Sum_Rape_Assault'] = pivot_df['Rape'].fillna(0) + pivot_df['Sexual assault'].fillna(0)
    pivot_df['Difference'] = pivot_df['Sexual violence'] - pivot_df['Sum_Rape_Assault']

# Display samples where we have data for all three categories to analyze alignment
display(pivot_df.dropna(subset=['Sexual violence', 'Rape', 'Sexual assault']).head(15))
```

```python
# Calculate how often Rape + Sexual assault perfectly equals Sexual violence
zero_diff_count = (pivot_df['Difference'] == 0).sum()
total_rows = pivot_df.dropna(subset=['Sexual violence', 'Rape', 'Sexual assault']).shape[0]

print(f"Total aligned records with all 3 categories present: {total_rows}")
print(f"Records where Rape + Sexual assault perfectly equals Sexual violence: {zero_diff_count}")

# Let's inspect a few rows where they don't match, if any
not_matching = pivot_df[pivot_df['Difference'] != 0].dropna(subset=['Sexual violence', 'Rape', 'Sexual assault'])
if len(not_matching) > 0:
    print("\nSample records where Rape + Sexual assault does NOT equal Sexual violence:")
    display(not_matching.head(10))
else:
    print("\nIn all available instances, Rape + Sexual assault perfectly equals Sexual violence.")
```

```python
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Load the rates dataset
df_rates = pd.read_csv('/content/crime_data_rates.csv')

# Filter for years 2020 to 2024 and the specified crimes
target_crimes = ['Intentional homicide', 'Sexual violence', 'Theft']
df_filtered = df_rates[
    (df_rates['time_period'].between(2020, 2024)) &
    (df_rates['international_classification_of_crime_for_statistical_purposes_iccs'].isin(target_crimes))
]

# Aggregate the average rate per country for each of these crimes over the 2020-2024 period
pivot_rates = df_filtered.pivot_table(
    index='geopolitical_entity_reporting',
    columns='international_classification_of_crime_for_statistical_purposes_iccs',
    values='obs_value',
    aggfunc='mean'
).dropna() # drop countries missing any of the 3 key features

# Standardize and scale log-transformed rates as requested
crimes = ['Intentional homicide', 'Sexual violence', 'Theft']
z = pd.DataFrame(StandardScaler().fit_transform(np.log1p(pivot_rates[crimes])),
                 index=pivot_rates.index, columns=crimes)

# Apply the custom weights to the standardized features
weights = {'Intentional homicide': 0.5, 'Sexual violence': 0.3, 'Theft': 0.1}
pivot_rates['Weighted_Crime_Score'] = sum(z[c] * w for c, w in weights.items())

# Perform 3-class K-Means clustering to classify countries into High, Medium, and Low crime categories
features = pivot_rates[['Weighted_Crime_Score']].values
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
pivot_rates['Cluster'] = kmeans.fit_predict(features)

# Determine the order of clusters based on their average score to map them correctly to Low, Medium, High
cluster_means = pivot_rates.groupby('Cluster')['Weighted_Crime_Score'].mean().sort_values()
cluster_mapping = {
    cluster_means.index[0]: 'Low Crime',
    cluster_means.index[1]: 'Medium Crime',
    cluster_means.index[2]: 'High Crime'
}
pivot_rates['Crime_Category'] = pivot_rates['Cluster'].map(cluster_mapping)

# Display the sorted classified list
classification_result = pivot_rates[['Intentional homicide', 'Sexual violence', 'Theft', 'Weighted_Crime_Score', 'Crime_Category']].sort_values(by='Weighted_Crime_Score')
display(classification_result)
classification_result.to_csv('/content/classification_result.csv')
```
